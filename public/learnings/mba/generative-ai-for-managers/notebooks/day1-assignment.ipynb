````
import os
import json
import time
import re
from datetime import datetime, timezone
from google import genai
from google.genai import types

# --- Load API key ---
try:
    from google.colab import userdata
    os.environ["GEMINI_API_KEY"] = userdata.get("GEMINI_API_KEY")
except Exception:
    pass

if not os.environ.get("GEMINI_API_KEY"):
    import getpass
    os.environ["GEMINI_API_KEY"] = getpass.getpass("Paste your GEMINI_API_KEY (input hidden): ")

# Initialize client
client = genai.Client()
MODEL_ID = "gemini-2.0-flash-lite"

# --- Prompt Logging ---
PROMPT_LOG = []

def _now():
    return datetime.now(timezone.utc).isoformat().replace('+00:00', 'Z')

def generate(prompt, temperature=0.7, max_tokens=500, log=True):
    """Generate text using Gemini with logging."""
    t0 = time.time()
    response = client.models.generate_content(
        model=MODEL_ID,
        contents=prompt,
        config=types.GenerateContentConfig(
            temperature=temperature,
            max_output_tokens=max_tokens
        )
    )
    text = response.text or ""
    latency = time.time() - t0

    if log:
        PROMPT_LOG.append({
            "ts": _now(),
            "temperature": temperature,
            "prompt": prompt,
            "output": text,
            "latency_s": round(latency, 3)
        })
    return text

def try_parse_json(text):
    """Attempt to parse JSON from LLM output."""
    try:
        return True, json.loads(text)
    except json.JSONDecodeError:
        pass
    json_match = re.search(r'```(?:json)?\s*([\s\S]*?)```', text)
    if json_match:
        try:
            return True, json.loads(json_match.group(1))
        except json.JSONDecodeError:
            pass
    json_match = re.search(r'[\[\{][\s\S]*[\]\}]', text)
    if json_match:
        try:
            return True, json.loads(json_match.group())
        except json.JSONDecodeError:
            pass
    return False, None

print("✅ Setup complete!")
````

## Sample A: Customer Review Response Generator

````
def generate_review_response(review_text, rating, product_name, store_name="RetailMax"):
    """
    Generate a professional response to a customer review.

    Args:
        review_text (str): The customer's review
        rating (int): Star rating (1-5)
        product_name (str): Name of the product reviewed
        store_name (str): Name of the store

    Returns:
        str: Professional response to the review
    """
    tone_guide = {
        5: "enthusiastic and grateful",
        4: "appreciative and warm",
        3: "understanding and helpful",
        2: "empathetic and solution-focused",
        1: "deeply apologetic and action-oriented"
    }

    tone = tone_guide.get(rating, tone_guide[3])

    prompt = f"""You are a customer service representative for {store_name}.
Write a response to this customer review.

Product: {product_name}
Rating: {rating}/5 stars
Review: {review_text}

Be {tone}.

Rules:
- Thank them for feedback
- Reference specific points from their review (do not be generic)
- If rating <= 3, offer a specific solution or next step
- Keep under 100 words
- Do not invent facts not in the review
- Sign off as "The {store_name} Team"

Response:"""

    return generate(prompt, temperature=0.4)


# Test
print("=== 5-STAR ===")
print(generate_review_response(
    "Amazing headphones! Battery lasts forever!",
    5, "SoundWave Pro"
))

print("\n=== 2-STAR ===")
print(generate_review_response(
    "Stopped working after 2 weeks. Very disappointed.",
    2, "SoundWave Pro"
))
````

## Sample B: Product Description Writer

````
def generate_product_description(product_name, specifications, target_audience):
    """
    Generate a product description from specifications.

    Args:
        product_name (str): Name of the product
        specifications (dict): Product specs
        target_audience (str): Who it's for

    Returns:
        str: Compelling product description
    """
    specs_text = "\n".join([f"- {k}: {v}" for k, v in specifications.items()])

    prompt = f"""Write a product description for an e-commerce website.

Product: {product_name}

Specifications:
{specs_text}

Target Audience: {target_audience}

Rules:
- Engaging opening (no generic "Introducing...")
- Translate specs into customer benefits
- Include 2-3 use cases
- Only mention features from the specifications above
- Do not invent features not listed
- 100-150 words

Description:"""

    return generate(prompt, temperature=0.6)


# Test
specs = {
    "Weight": "250g",
    "Battery": "40 hours",
    "Connectivity": "Bluetooth 5.0",
    "ANC": "Hybrid with transparency mode"
}

print(generate_product_description(
    "SoundWave Pro Headphones",
    specs,
    "Remote professionals"
))
````

## Sample C: Internal Memo Summarizer

````
def summarize_memo(memo_text, summary_format="executive"):
    """
    Summarize an internal company memo.

    Args:
        memo_text (str): Full memo
        summary_format (str): "executive", "action", or "team"

    Returns:
        str: Formatted summary
    """
    formats = {
        "executive": "3-sentence summary: key decision, impact, timeline.",
        "action": "Numbered list of action items with owners and deadlines.",
        "team": "TL;DR (1 sentence), Key Points (bullets), Action Items."
    }

    prompt = f"""Summarize this internal memo.

Memo:
{memo_text}

Format: {formats.get(summary_format, formats['executive'])}

Rules:
- Only include information from the memo
- If owner/deadline not mentioned, write "Not specified"
- Do not invent facts

Summary:"""

    return generate(prompt, temperature=0.2)


# Test
memo = """
TO: All Department Heads
FROM: CEO
RE: Q1 2025 Initiatives

We're increasing digital investment by 25% ($2M to IT).
Sarah Chen leads this with Deloitte support.

New CRM training for all stores by March 15.
Regional managers: submit schedules to HR by Jan 25.

Travel freeze until Q2. Requests over $500 need VP approval.

3% cost-of-living raise effective Feb 1.

Q&A session: Jan 22 at 2pm.
"""

print("=== EXECUTIVE ===")
print(summarize_memo(memo, "executive"))

print("\n=== ACTIONS ===")
print(summarize_memo(memo, "action"))
````

## Sample D: Job Posting Generator

````
def generate_job_posting(job_title, department, requirements, nice_to_have=None):
    """
    Generate a job posting from requirements.

    Args:
        job_title (str): Position title
        department (str): Department
        requirements (list): Required qualifications
        nice_to_have (list): Optional qualifications

    Returns:
        str: Complete job posting
    """
    reqs = "\n".join([f"- {r}" for r in requirements])
    nice = "\n".join([f"- {n}" for n in (nice_to_have or [])])

    prompt = f"""Write a job posting for RetailMax, a mid-size retail company.

Position: {job_title}
Department: {department}

Required Qualifications:
{reqs}

{"Nice to Have:" if nice_to_have else ""}
{nice if nice_to_have else ""}

Include these sections:
1. Engaging opening paragraph about the role
2. "What You'll Do" (5 responsibilities based on the role)
3. "What You'll Bring" (from requirements above)
4. "Why Join Us" (standard retail company benefits)

Rules:
- Only use requirements listed above
- Do not add qualifications not provided
- 300-400 words
- Professional but approachable tone

Job Posting:"""

    return generate(prompt, temperature=0.5)


# Test
reqs = [
    "3+ years retail management",
    "Strong leadership skills",
    "Inventory management experience",
    "Flexible schedule including weekends"
]

nice = [
    "Bachelor's in Business",
    "CRM experience",
    "Bilingual English/Spanish"
]

print(generate_job_posting("Store Manager", "Retail Operations", reqs, nice))
````
